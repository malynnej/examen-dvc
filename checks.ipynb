```python
import pandas as pd
from pathlib import Path

# Adjust these paths/names to your project layout
data_dir = Path("data/processed_data")

# Load processed datasets (CSV or Parquet – pick the one you use)
X_train = pd.read_parquet(data_dir / "X_train.parquet")
X_test  = pd.read_parquet(data_dir / "X_test.parquet")
X_train_scaled = pd.read_parquet(data_dir / "X_train_scaled.parquet")
X_test_scaled  = pd.read_parquet(data_dir / "X_test_scaled.parquet")
y_train = pd.read_parquet(data_dir / "y_train.parquet").squeeze()
y_test  = pd.read_parquet(data_dir / "y_test.parquet").squeeze()

print("Shapes:")
print(f"X_train: {X_train.shape}")
print(f"X_test : {X_test.shape}")
print(f"X_train scaled: {X_train_scaled.shape}")
print(f"X_test scaled: {X_test_scaled.shape}")
print(f"y_train: {y_train.shape}")
print(f"y_test : {y_test.shape}")

print("\nDtypes X_train:")
print(X_train.dtypes)

# Drop non-numeric columns for numeric stats (e.g. date)
X_train_num = X_train.select_dtypes(include=["float64", "int64"])
X_test_num  = X_test.select_dtypes(include=["float64", "int64"])
X_train_scaled_num= X_train_scaled.select_dtypes(include=["float64", "int64"])
X_test_scaled_num  = X_test_scaled.select_dtypes(include=["float64", "int64"])

print("\nX_train numeric describe():")
display(X_train_num.describe().T)  # .T = nicer row-wise view

print("\nX_test numeric describe():")
display(X_test_num.describe().T)

print("\nX_train scaled numeric describe():")
display(X_train_scaled_num.describe().T)  # .T = nicer row-wise view

print("\nX_test scaled numeric describe():")
display(X_test_scaled_num.describe().T)

print("\ny_train describe():")
display(y_train.describe())

print("\ny_test describe():")
display(y_test.describe())

# Optional: compare train/test means for each feature
train_means = X_train_num.mean().rename("train_mean")
test_means  = X_test_num.mean().rename("test_mean")
mean_compare = pd.concat([train_means, test_means], axis=1)
mean_compare["abs_diff"] = (mean_compare["train_mean"] - mean_compare["test_mean"]).abs()

print("\nFeature-wise mean comparison (train vs test):")
display(mean_compare)

# Optional: compare train/test means for each feature
train_scaled_means = X_train_scaled_num.mean().rename("train_mean")
test_scaled_means  = X_test_scaled_num.mean().rename("test_mean")
mean_scaled_compare = pd.concat([train_scaled_means, test_scaled_means], axis=1)
mean_scaled_compare["abs_diff"] = (mean_compare["train_mean"] - mean_compare["test_mean"]).abs()

print("\nFeature-wise mean comparison (scaled train vs test):")
display(mean_scaled_compare)
```
